# import necessary libraries

```python
import numpy as np
import pandas as pd
import warnings
warnings.filterwarnings('ignore')
```

# Import diabetes.csv data ser using sklearn

```python
df = pd.read_csv('diabetes.csv')
```

```python
df
```

# Check the distribution of the target

```python
df['Outcome'].value_counts()
```

# Replace zeros with NaN for specific columns

```python
#Count how many zero values each column has
(df == 0).sum()
```

```python
#Pregnancy is allowed to be 0
cols_with_zero = ['Glucose', 'BloodPressure', 'SkinThickness', 'Insulin', 'BMI']
```

```python
df[cols_with_zero] = df[cols_with_zero].replace(0, np.nan)
```

```python
(df == 0).sum()
```

# Check for missing values

```python
#Verify the zeros were converted to nan
df.isnull().sum()
```

# Fill missing values with median

```python
#Fill missing values in each column with that column's median
for c in cols_with_zero:
    df[c] = df[c].fillna(df[c].median())
```

```python
df.isnull().sum()
```

# Visualize Distributions

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df.hist(figsize = (12, 10), bins = 10)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize = (10, 8), dpi = 2500)
sns.heatmap(df.corr(numeric_only = True), annot = True, cmap = 'viridis', fmt = '.2f')
# plt.savefig('Heatmap.png', dpi = 5000)
```

```python
df.plot(kind = 'box', subplots = True, layout = (3, 3), figsize = (14, 10))
plt.tight_layout()
plt.show()
```

# Convert Targer data into interger code

```python
df['Outcome'] = df['Outcome'].map({'Diabetic': 1, 'Non Diabetic': 0})
```

```python
df
```

# Divide the data into input and output

```python
from sklearn.model_selection import train_test_split
x = df.drop('Outcome', axis = 1)
```

```python
y = df['Outcome']
```

# Splitting the dataset into the Training set and Test set

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.5, random_state = 42)
```

# Create Model

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
model = DecisionTreeClassifier(max_depth = 4)
```

# Fitting DecisionTreeClassifier   on dataset

```python
model.fit(x_train, y_train)
```

# Display Decision Tree

```python
from sklearn.tree import plot_tree
```

```python
plt.figure(figsize = (20, 15))
plot_tree(model,
         feature_names = x.columns,
         class_names = ['Non Diabetic', 'Diabetic'],
         filled = True, rounded = True)
plt.show()
```

# Predict the x_test 

```python
y_pred = model.predict(x_test)
y_pred
```

# Display Training Accuracy

```python
model.score(x_train,y_train)
```

# Display Test Accuracy

```python
model.score(x_test,y_test)
```

```python
from sklearn.metrics import accuracy_score
```

```python
display(accuracy_score(y_test, y_pred))
```

```python
from sklearn.metrics import confusion_matrix, classification_report
```

```python
print('Confusion Matrix: ')
print(confusion_matrix(y_test, y_pred))
```

```python
print('Classification Report: ')
print(classification_report(y_test, y_pred))
```
